```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("adamgbor/club-football-match-data-2000-2025")

print("Path to dataset files:", path)
```

```python
# Initial data preprocessing
Raw = pd.read_csv(f"{path}/Matches.csv")

display(Raw.columns.tolist())

display(Raw.head())
```

```python
# Dropping unneeded features

SelectedFeatures = Raw.drop(['Division',
                            'MatchTime',
                            'FTHome',
                            'FTAway',
                            'HTResult',
                            'HomeShots',
                            'AwayShots',
                            'HomeTarget',
                            'AwayTarget',
                            'HomeFouls',
                            'AwayFouls',
                            'HomeCorners',
                            'AwayCorners',
                            'HomeYellow',
                            'AwayYellow',
                            'HomeRed',
                            'AwayRed',
                            'MaxHome',
                            'MaxDraw',
                            'MaxAway',
                            'Over25',
                            'Under25',
                            'MaxOver25',
                            'MaxUnder25',
                            'HandiSize',
                            'HandiHome',
                            'HandiAway',
                            'C_LTH',
                            'C_LTA',
                            'C_VHD',
                            'C_VAD',
                            'C_HTB',
                            'C_PHB'], axis=1)

display(SelectedFeatures.head())
print("\n")


# Creating a new dataframe for the games that Aston Villa has played

from src.tools import filterTeam

VillaGames = filterTeam(SelectedFeatures, "Aston Villa")

display(VillaGames.head())
display(VillaGames.shape)
print("\n")


# Saving the games Aston Villa has played to drive

os.makedirs(f"{DATA}/processed", exist_ok=True)
VillaGames.to_csv(f"{DATA}/processed/villa_matches.csv", index=False)
print("Saved:", os.listdir(f"{DATA}/processed"))
print("\n")

```
