```python
# Initial data preprocessing
Raw = pd.read_csv("/kaggle/input/club-football-match-data-2000-2025/Matches.csv")

display(Raw.columns.tolist())

display(Raw.head())

Raw.shape
```

```python
# Dropping unneeded features

SelectedFeatures = Raw.drop(['Division',
                            'MatchDate',
                            'MatchTime',
                            'FTHome',
                            'FTAway',
                            'HTResult',
                            'HomeShots',
                            'AwayShots',
                            'HomeTarget',
                            'AwayTarget',
                            'HomeFouls',
                            'AwayFouls',
                            'HomeCorners',
                            'AwayCorners',
                            'HomeYellow',
                            'AwayYellow',
                            'HomeRed',
                            'AwayRed',
                            'MaxHome',
                            'MaxDraw',
                            'MaxAway',
                            'Over25',
                            'Under25',
                            'MaxOver25',
                            'MaxUnder25',
                            'HandiSize',
                            'HandiHome',
                            'HandiAway',
                            'C_LTH',
                            'C_LTA',
                            'C_VHD',
                            'C_VAD',
                            'C_HTB',
                            'C_PHB'], axis=1)

display(SelectedFeatures.head())
print("\n")


# Filtering only the games that Aston Villa has played

VillaGames = SelectedFeatures.loc[(SelectedFeatures['HomeTeam']=="Aston Villa") |
                                  (SelectedFeatures['AwayTeam']=="Aston Villa")].copy()

display(VillaGames.head())
display(VillaGames.shape)
print("\n")

''' processedData = pd.DataFrame(columns=['TeamElo', 'OpponentElo', 'Form3Team',
                                      'Form3Opponent', 'Form5Team',
                                      'Form5Opponent', 'HTTeam', 'HTOpponent',
                                      'OddTeam', 'OddOpponent', 'OddDraw',
                                      'Result'])
'''
```
